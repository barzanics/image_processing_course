```python
import sys                                    # lets us change where Python looks for files
sys.path.append("../..")                      # utils.py and matcher.py live two folders up
import cv2                                    # OpenCV
from utils import load, show, show_many       # our helpers
print("Libraries loaded - ready to go!")      # a visible sign that everything worked
```

### Task 1 - Which is brighter in gray: the centre of the first red circle (row 90, col 100) or of the first green square (row 210, col 105)?

```python
img = load("../../assets/objects_shapes.png")              # the coloured shapes
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)               # to gray
red_value, green_value = gray[90, 100], gray[210, 105]     # the two pixels
print("red circle:", red_value, "| green square:", green_value)
print("brighter one:", "green square" if green_value > red_value else "red circle")
```

### Task 2 - Show only the blue channel as a gray picture and print its value inside a blue triangle (row 350, col 110) and inside a red circle (row 90, col 100).

```python
blue, green, red = cv2.split(img)                          # three channels
print("blue channel inside the triangle:", blue[350, 110], "| inside the red circle:", blue[90, 100])
show(blue, "blue channel: triangles are bright", size=(6, 4))
```

### Task 3 - Make the picture 50% darker and print its brightest value.

```python
darker = cv2.convertScaleAbs(img, alpha=0.5, beta=0)       # multiply every number by 0.5
print("brightest value now:", darker.max(), "(255 * 0.5 = 127.5, rounded)")
show_many([img, darker], ["original", "50% darker"], columns=2, size=(10, 3))
```

### Task 4 - Make a thumbnail that is 100 pixels wide and keeps the proportions.

```python
height, width = img.shape[0], img.shape[1]                 # 420 x 600
new_height = int(height * 100 / width)                     # keep the same proportions
thumb = cv2.resize(img, (100, new_height))                 # (width, height)!
print("thumbnail shape:", thumb.shape)
```
