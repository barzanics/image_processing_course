**Task:** the teacher gives you `assets/incoming/myimage.png`. Extract and print every number in it.

Below `IMAGE_PATH` points to the ready-made practice picture `sample_numbers.png`. **To use the teacher's picture, change that one line to `"../../assets/incoming/myimage.png"`.** Remember the limits from Notebook 08: the picture must have dark digits on a lighter background, upright and in a plain font.

```python
import matcher                                            # our finished toolbox
IMAGE_PATH = "../../assets/incoming/sample_numbers.png"   # <- change to myimage.png
img = load(IMAGE_PATH)                                    # colour picture
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)              # gray
print("picture shape:", img.shape)
show(img, "the picture", size=(8, 4))
```

Now the whole capstone in one call, then print every number with its row.

```python
rows = matcher.read_all_numbers(gray)                     # rows of numbers
all_numbers = []
for row_number in range(len(rows)):
    for number in rows[row_number]:
        print("row", row_number + 1, ":", int(number))
        all_numbers.append(int(number))
print("every number:", all_numbers)
```

Finally draw the result: a green box around every digit (the same boxes the reader used).

```python
closed, boxes = matcher.find_boxes_adaptive(gray)
result = img.copy()
for x, y, w, h in boxes:
    cv2.rectangle(result, (x - 3, y - 3), (x + w + 3, y + h + 3), (0, 200, 0), 2)
print("digits boxed:", len(boxes))
show(result, "every digit found", size=(8, 4))
```
