### Task 1 - Print only the numbers with 4 digits found in `digits_grid_photo_like.png`.

```python
import matcher                                            # our finished toolbox
gray = cv2.cvtColor(load("../../assets/digits_grid_photo_like.png"), cv2.COLOR_BGR2GRAY)
rows = matcher.read_all_numbers(gray)                      # [['52', '7319'], ['908'], ...]
four_digit = []
for row in rows:
    for number in row:
        if len(number) == 4:
            four_digit.append(int(number))
print("4-digit numbers:", four_digit)
```

### Task 2 - Add up every number in the picture.

```python
total = 0
for row in rows:
    for number in row:
        total = total + int(number)
print("sum of all numbers:", total)
```

### Task 3 - Save a crop of the biggest digit (largest box area) to `assets/out/biggest_digit.png`.

```python
import os
closed, boxes = matcher.find_boxes_adaptive(gray)          # white digits on black + boxes
best = boxes[0]
for box in boxes:
    if box[2] * box[3] > best[2] * best[3]:                # width * height
        best = box
x, y, w, h = best
os.makedirs("../../assets/out", exist_ok=True)
cv2.imwrite("../../assets/out/biggest_digit.png", closed[y:y + h, x:x + w])
print("biggest digit box:", best, "| saved:", os.path.exists("../../assets/out/biggest_digit.png"))
```
