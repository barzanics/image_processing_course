### Task 1 - Print the size of `letters_clean.png` as: "The picture is W pixels wide and H pixels tall."

### Task 2 - Print the B, G, R values of `objects_shapes.png` at row 350, column 110 and name the colour.

```python
img = load("../../assets/objects_shapes.png")              # the coloured shapes
pixel = img[350, 110]                                      # row first, then column
print("BGR =", pixel, "-> the blue number is the largest, so it is a BLUE triangle")
```

### Task 3 - Crop the middle green square (rows 185-235, columns 250-300), show it and print its shape.

```python
square = img[185:235, 250:300]                             # rows, then columns
print("crop shape:", square.shape)
show(square, "the middle green square", size=(3, 3))
```

### Task 4 - On a copy, paint a 20 x 20 red square in the top-left corner and show it.

```python
edited = img.copy()                                        # never paint on the original
edited[0:20, 0:20] = (0, 0, 255)                           # BGR red
print("pixel (5, 5) is now", edited[5, 5], "| original still", img[5, 5])
show(edited, "red square in the corner", size=(5, 4))
```
