### Task 1 - Count the contours in `digits_single.png` and print the width of each box.

```python
gray = cv2.cvtColor(load("../../assets/digits_single.png"), cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)  # white digits
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
widths = []
for contour in contours:
    x, y, w, h = cv2.boundingRect(contour)
    widths.append(w)
print("contours:", len(contours), "| widths:", sorted(widths))
```

### Task 2 - In `letters_noisy.png`, how many contours have an area below 60 (the specks)?

```python
noisy = cv2.cvtColor(load("../../assets/letters_noisy.png"), cv2.COLOR_BGR2GRAY)
blur = cv2.medianBlur(noisy, 5)
_, otsu = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
contours, _ = cv2.findContours(cv2.bitwise_not(otsu), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
specks = 0
for contour in contours:
    if cv2.contourArea(contour) < 60:
        specks = specks + 1
print("all contours:", len(contours), "| specks:", specks, "| letters:", len(contours) - specks)
```

### Task 3 - In `digits_number.png`, print the left edge (x) of each digit from left to right.

```python
gray = cv2.cvtColor(load("../../assets/digits_number.png"), cv2.COLOR_BGR2GRAY)
_, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
boxes = []
for contour in contours:
    boxes.append(cv2.boundingRect(contour))
boxes = sorted(boxes)                                      # left to right
left_edges = []
for box in boxes:
    left_edges.append(box[0])
print("left edges:", left_edges)
```
