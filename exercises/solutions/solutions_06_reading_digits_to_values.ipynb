### Task 1 - Read `digits_single.png` and print how many of the 10 digits are right.

```python
import matcher                                            # our finished toolbox
templates = matcher.make_templates(matcher.DIGITS)
gray = cv2.cvtColor(load("../../assets/digits_single.png"), cv2.COLOR_BGR2GRAY)
thresh, boxes = matcher.find_boxes(gray)
text = ""
for x, y, w, h in boxes:
    text = text + matcher.recognize(thresh[y:y + h, x:x + w], templates)[0]
print("I read:", text, "| correct:", text == "0123456789")
```

### Task 2 - Read `digits_number.png` and print the two numbers and their sum.

```python
gray = cv2.cvtColor(load("../../assets/digits_number.png"), cv2.COLOR_BGR2GRAY)
thresh, boxes = matcher.find_boxes(gray)
texts = matcher.read_row(thresh, boxes, templates)         # ['427', '1058']
print("numbers:", int(texts[0]), int(texts[1]), "| sum:", int(texts[0]) + int(texts[1]))
```

### Task 3 - Draw the number `90210` yourself and read it back as an integer.

```python
canvas = np.full((140, 420, 3), 255, dtype=np.uint8)
cv2.putText(canvas, "90210", (20, 105), cv2.FONT_HERSHEY_SIMPLEX, 3.0, (0, 0, 0), 5, cv2.LINE_AA)
thresh, boxes = matcher.find_boxes(cv2.cvtColor(canvas, cv2.COLOR_BGR2GRAY))
texts = matcher.read_row(thresh, boxes, templates)
print("integer:", int("".join(texts)), "| type:", type(int("".join(texts))).__name__)
```
