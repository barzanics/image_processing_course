```python
import sys                                    # lets us change where Python looks for files
sys.path.append("../..")                      # utils.py and matcher.py live two folders up
import cv2                                    # OpenCV
import numpy as np                            # NumPy
from utils import load, show, show_many       # our helpers
print("Libraries loaded - ready to go!")      # a visible sign that everything worked
```

### Task 1 - Apply Otsu to `letters_clean.png` and print the threshold it chose.

```python
gray = cv2.cvtColor(load("../../assets/letters_clean.png"), cv2.COLOR_BGR2GRAY)
value, binary = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print("Otsu chose:", value)
```

### Task 2 - Make white-text-on-black and count the white pixels.

```python
white_on_black = cv2.bitwise_not(binary)                   # text becomes 255
print("white (text) pixels:", cv2.countNonZero(white_on_black))
show(white_on_black, "white text on black", size=(8, 3))
```

### Task 3 - Clean `letters_noisy.png` with a median blur (5) and Otsu, then show the before and after.

```python
noisy = cv2.cvtColor(load("../../assets/letters_noisy.png"), cv2.COLOR_BGR2GRAY)
blur = cv2.medianBlur(noisy, 5)                            # remove grain
_, cleaned = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print("values left:", np.unique(cleaned))
show_many([noisy, cleaned], ["noisy", "median + Otsu"], columns=1, size=(8, 5))
```
