### Task 1 - Read `letters_clean.png` and print the weakest (lowest) letter confidence.

```python
import matcher                                            # our finished toolbox
templates = matcher.make_templates(matcher.ALPHABET)      # A-Z templates
gray = cv2.cvtColor(load("../../assets/letters_clean.png"), cv2.COLOR_BGR2GRAY)
thresh, boxes = matcher.find_boxes(gray)                   # white letters + sorted boxes
lowest = 1.0
for x, y, w, h in boxes:
    letter, score, second, second_score = matcher.recognize(thresh[y:y + h, x:x + w], templates)
    if score < lowest:
        lowest = score
print("weakest confidence:", round(lowest, 2))
```

### Task 2 - Draw the word `OPEN` with `cv2.putText` on a white picture and read it back.

```python
canvas = np.full((120, 300, 3), 255, dtype=np.uint8)       # white picture
cv2.putText(canvas, "OPEN", (20, 90), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (0, 0, 0), 4, cv2.LINE_AA)
thresh, boxes = matcher.find_boxes(cv2.cvtColor(canvas, cv2.COLOR_BGR2GRAY))
word = ""
for x, y, w, h in boxes:
    word = word + matcher.recognize(thresh[y:y + h, x:x + w], templates)[0]
print("I read:", word)
```

### Task 3 - Deskew `letters_skewed.png` (it is tilted by 10 degrees) and read it.
